```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

## Dataset Description

This task uses a sample employee dataset containing employee names, ages, salaries, and departments.

Some values are intentionally left missing to demonstrate missing-value detection and analysis.

The dataset is created using Pandas and NumPy and saved as a CSV file.

```python
data = {
    "Name": [
        "Arun", "Bala", "Charan", "Divya", "Ezhil",
        "Farhan", "Gokul", "Hari", "Indhu", "Kavin"
    ],
    "Age": [
        22, np.nan, 25, 23, np.nan,
        24, 21, np.nan, 26, 22
    ],
    "Salary": [
        25000, 30000, np.nan, 28000, 32000,
        np.nan, 27000, 29000, np.nan, 31000
    ],
    "Department": [
        "ECE", "CSE", "ECE", np.nan, "IT",
        "CSE", np.nan, "IT", "ECE", "CSE"
    ]
}

df = pd.DataFrame(data)

df.to_csv("dataset.csv", index=False)

print("Dataset created and saved successfully!")

df
```

## Load the Dataset

The CSV file is loaded into a Pandas DataFrame using `read_csv()`.

The shape and first few rows are checked to understand the dataset structure.

```python
df = pd.read_csv("dataset.csv")

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head(10)
```

## Dataset Information

The `info()` method displays column names, data types, and the number of non-null values. This helps identify columns that may contain missing data.

```python
df.info()
```

## Identify Missing Values

The `isnull()` method identifies missing values in the dataset. The `sum()` method counts missing values column by column.

```python
missing_counts = df.isnull().sum()

print("Missing values in each column:")
print(missing_counts)
```

## Total Missing Values

The total number of missing values is calculated by summing the column-wise missing-value counts.

```python
total_missing = df.isnull().sum().sum()

print("Total number of missing values:", total_missing)
```

## Missing-Value Percentage

The percentage of missing values in each column is calculated by dividing its missing-value count by the total number of rows and multiplying by 100.

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing-value percentage in each column:")
print(missing_percentage.round(2))
```

## Missing-Value Summary

A summary table is created to combine the number and percentage of missing values for every column. This makes it easier to identify incomplete columns.

```python
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_summary
```

## Identify Columns Requiring Treatment

Columns containing one or more missing values are identified for further investigation.

Possible treatments include filling missing numerical values with the median and categorical values with the mode. The appropriate method depends on the dataset and the intended analysis.

```python
columns_requiring_treatment = missing_summary[
    missing_summary["Missing Count"] > 0
]

print("Columns requiring treatment:")
display(columns_requiring_treatment)
```

## Visualize Missing Values

A bar chart is generated to compare the number of missing values across columns. Visualization makes it easier to identify columns with more incomplete data.

```python
missing_counts = df.isnull().sum()

missing_counts.plot(
    kind="bar",
    figsize=(8, 5),
    title="Missing Values by Column"
)

plt.xlabel("Column")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Key Observations

1. The dataset contains 10 rows and 4 columns.
2. A total of 8 missing values were identified.
3. The Age column contains 3 missing values (30%).
4. The Salary column contains 3 missing values (30%).
5. The Department column contains 2 missing values (20%).
6. The Name column contains no missing values.
7. Age, Salary, and Department require further investigation and appropriate missing-value treatment.
